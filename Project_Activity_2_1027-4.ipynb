## Introduction

My main research question is: **How have abortion provision rates in the United States changed since the overturning of Roe v. Wade in June 2022?** In this project I examined the changes in state provision rates, the relationship between state poverty and abortion provision, the geographic distribution of abortions, and the effects on interstate travel to access abortion-related healthcare. The project describes patterns and associations rather than estimating the nationwide causal effect of abortion restrictions as those can't be determined without taking into account socioeconomic, public health, sociopolitical, and environmental issues.

The analysis uses Guttmacher’s 2020 Abortion Provider Census findings with its Monthly Abortion Provision Study and interstate travel estimates. The historical table contains state abortion counts and rates for 2017, 2019, and 2020; I use 2020 as the principal pre-Dobbs baseline. Post-Dobbs analyses focus on the years 2023 and 2024. Census population estimates provide the number of resident women aged 15–44 used to calculate post-Dobbs rates. Census poverty data provide official state poverty percentages pooled across 2023–2025. The main variables I used are as follows: state, month, estimated abortion count, abortion provision rate, poverty percentage, and residents traveling out of state for abortion care. These sources provide state-level information suitable to examine provision and travel, although they do not directly measure unmet need, individual expenses, or delays in obtaining care.

## Data Analysis and Results

### Data Preparation

I used pandas to inspect data types, missing values, and duplicate observations. I converted the monthly abortion dates to datetime values and matched state abbreviations with full state names to account for differences in how they were presented in all my different datasets. I filtered the Census population records to include females aged 15–44, selecting all Hispanic origins together and summing across ages and race categories. I aggregated monthly abortion estimates into annual state counts, calculating rates per 1,000 resident women aged 15–44. Some states had only six months of observations in 2023, so their annual counts and rates were left missing, I did not treat the partial totals as full years to be as accurate as possible. All states and DC had 12 months of observations in 2024. I removed spreadsheet headings and footnotes from the poverty data and merged the relevant percentages by state to examine the relationships between the two variables. I used the published out-of-state summary row to investigate travel.

### Changes in State Abortion Provision Rates

The state comparison found that 37 states had higher abortion provision rates in 2024 than in 2020, while 13 had lower rates. The largest increases were in New Mexico, Kansas, and Illinois: approximately 29.41, 24.17, and 15.71 additional abortions were recorded per 1,000 women aged 15–44, respectively. The largest decreases were in Oklahoma and North Dakota, with approximately 8.80 and 6.05 fewer abortions per 1,000 women aged 15–44. The state-change chart shows how differently provision rates changed across states. These changes reflect differences between near total-ban states post the Dobbs ruling and states with expanded access.

I conducted a supplementary comparison using the same 39 states with complete annual rates in both 2023 and 2024, alongside their 2020 rates. The boxplot diagram shows the distribution of rates within this consistent group. The states with incomplete 2023 observations are excluded, hence this comparison does not represent all states. 

### State Poverty and Abortion Provision

The correlation between pooled 2023–2025 official state poverty percentages collected from census data and 2024 abortion provision rates was −0.028. The poverty scatterplot therefore provides little evidence of a linear relationship between these two state-level variables. 
This result however, does not establish that poverty is unrelated to an individual's ability to obtain abortions. 

However, this does not mean poverty is unrelated to a person’s ability to obtain an abortion. The provision figures include patients visiting from other states, and their income level is unaccounted for in this data. We can still say that there is little evidence to establish a meaningful, linear relationship between the two variables based on the data we have.

### Interstate Travel and Statistical Inference

I calculated interstate travel rates per 1,000 resident women aged 15–44 and compared 2023 with 2024. The paired analysis included 47 states with available travel totals in both years. I used a two-sided paired t-test because I compared each state with itself.

Let $\mu_d$ represent the mean state change in travel rates, calculated as 2024 minus 2023:

- **Null hypothesis:** The mean state interstate abortion travel rate did not change between 2023 and 2024: $H_0: \mu_d = 0$.
- **Alternative hypothesis:** The mean state interstate abortion travel rate changed between 2023 and 2024: $H_A: \mu_d \ne 0$.
- **Significance level:** $\alpha = 0.05$.

The mean state travel rate decreased from 3.139 to 2.794, a mean change of −0.345 travelers per 1,000 women aged 15–44. The test returned **$t(46) = -2.394$** and **$p = 0.0208$**. The p-value was below 0.05, so I rejected the null hypothesis under the test’s assumptions.

### Florida Case Study

Florida illustrates how an individual state’s experience can differ from the broader average. Florida residents traveling out of state increased from 2,770 in 2023 to 8,210 in 2024—an increase of 5,440, or 196.4%.

The monthly provision analysis showed a change around May 2024 as well. During January–April, the average monthly abortion provision was 17.06% higher in 2024 than across January-April in 2023. During May–December, average monthly abortion provision was 28.07% lower, falling from around 7,249 to 5,214 abortions per month. According to Guttmacher Institute (2024), Florida’s six-week ban took effect on May 1, 2024, which suggest the drops in abortion provision and increase in travel across state-lines may be linked.

## Conclusion and Future Work

Abortion provision did not change uniformly across states after Dobbs. Most states recorded higher provision rates in 2024 than in 2020, but there are also several states that experienced decreases. State poverty showed almost no linear association with provision rates from these datasets. Interstate travel rates decreased on average among the 47 included states, but there was a large increase in interstate travel in Florida alongside declining monthly abortion provision after May, 2024. 

This project can provide a general analysis on trends depicted in the context of abortion provision and abortion related travel throughout the United States after the Dobbs ruling, but causal effects and direct correlations cannot be determined and it cannot measure the difficulties individuals may face when it comes to obtaining abortion-related healthcare. Future research could separate in-person provision from telehealth, incorporate historical policy effective dates, and examine patients’ costs, delays, and unmet need. Future projects could also examine the relationship between individual poverty and follow-throughs after seeking abortions to better examine the relationship between abortion access and poverty. These additions would connect changes in recorded provision more directly to individuals’ access to care.

### Annual Abortion Counts and Rates

```python
# Sum estimated monthly abortion counts to obtain annual 2024 state totals
annual_2024 = (abortion_2024.groupby("state")["median"].sum().reset_index(name="abortions_2024"))
annual_2024["state_name"] = annual_2024["state"].map(state_names)
annual_2024.head(50)
```

```python
# Leave out annual counts when fewer than 12 months of data is available
annual_2023 = (
    abortion_2023.groupby("state")
    .agg(
        abortions_2023=("median", "sum"),
        months_available=("month", "nunique"),
        counts_available=("median", "count")
    )
    .reset_index()
)

# Mark incomplete annual counts as missing
annual_2023.loc[
    (annual_2023["months_available"] != 12) |
    (annual_2023["counts_available"] != 12),
    "abortions_2023"
] = np.nan

annual_2023["state_name"] = annual_2023["state"].map(state_names)

annual_2023.head(50)
```

```python
abortion_rates_2023 = annual_2023.merge(population_23, on="state_name", how="left", validate="one_to_one")

print("Missing population:",
      abortion_rates_2023["women_15_44_2023"].isna().sum())

print("Zero or negative population:",
      (abortion_rates_2023["women_15_44_2023"] <= 0).sum())
```

```python
abortion_rates_2024 = annual_2024.merge(population_24, on="state_name", how="left", validate="one_to_one")

print("Missing population:",
      abortion_rates_2024["women_15_44_2024"].isna().sum())

print("Zero or negative population:",
      (abortion_rates_2024["women_15_44_2024"] <= 0).sum())
```

```python
# Calculate abortions per 1,000 resident women aged 15–44.
abortion_rates_2023["abortion_rate_2023"] = (
    abortion_rates_2023["abortions_2023"] /
    abortion_rates_2023["women_15_44_2023"]
) * 1000

abortion_rates_2023 = abortion_rates_2023[
    [
        "state",
        "state_name",
        "abortions_2023",
        "women_15_44_2023",
        "abortion_rate_2023"
    ]
]

abortion_rates_2023.head(25)
```

## Changes in State Abortion Provision Rates

```python
# Match state rates across years & Calculate the change between abortion provision from 2020 to 2024.
before_states = abortion_before[
    abortion_before["geography"].isin(state_names.values())
][
    [
        "geography",
        "abortion_rate_2017",
        "abortion_rate_2019",
        "abortion_rate_2020"
    ]
].copy()

before_states = before_states.rename(
    columns={"geography": "state_name"}
)

rate_comparison = before_states.merge(
    abortion_rates_2023[
        ["state_name", "abortion_rate_2023"]
    ],
    on="state_name",
    how="left",
    validate="one_to_one"
)

rate_comparison = rate_comparison.merge(
    abortion_rates_2024[
        ["state_name", "abortion_rate_2024"]
    ],
    on="state_name",
    how="left",
    validate="one_to_one"
)

rate_comparison = rate_comparison[
    rate_comparison["state_name"] != "District of Columbia"
].copy()

rate_comparison["change_2020_2024"] = (
    rate_comparison["abortion_rate_2024"] -
    rate_comparison["abortion_rate_2020"]
)

rate_comparison.head(10)
```

```python
print("States with an increase from 2020 to 2024:")
print((rate_comparison["change_2020_2024"] > 0).sum())

print("\nStates with a decrease from 2020 to 2024:")
print((rate_comparison["change_2020_2024"] < 0).sum())

print("\nLargest increases:")
print(
    rate_comparison[
        ["state_name", "change_2020_2024"]
    ].sort_values(
        "change_2020_2024",
        ascending=False
    ).head(5)
)

print("\nLargest decreases:")
print(
    rate_comparison[
        ["state_name", "change_2020_2024"]
    ].sort_values(
        "change_2020_2024"
    ).head(5)
)
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=rate_comparison,
    x="abortion_rate_2020",
    y="abortion_rate_2024"
)

maximum_rate = rate_comparison[
    ["abortion_rate_2020", "abortion_rate_2024"]
].max().max()

plt.plot(
    [0, maximum_rate],
    [0, maximum_rate],
    color="gray",
    linestyle="--",
    label="Same rate in both years"
)

plt.title("State Abortion Provision Rates: 2020 and 2024")
plt.xlabel("2020 abortions per 1,000 women aged 15–44")
plt.ylabel("2024 abortions per 1,000 women aged 15–44")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
state_changes = rate_comparison.sort_values(
    "change_2020_2024")

colors = ["steelblue" if change >= 0 else "darkorange"
    for change in state_changes["change_2020_2024"]]

plt.figure(figsize=(10, 14))

plt.barh(state_changes["state_name"], state_changes["change_2020_2024"], color=colors)

plt.axvline(0, color="black", linewidth=1)

plt.title("Change in State Abortion Provision Rates, 2020–2024")
plt.xlabel("Change in abortions per 1,000 women aged 15–44")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

### Travel Changes Across States

```python
residence_columns = [
    f"{state}_residence"
    for state in state_names
    if state != "DC"
]

travel_comparison = pd.DataFrame({
    "state": [
        column.replace("_residence", "")
        for column in residence_columns
    ]
})

for year, data in [(2023, travel_2023), (2024, travel_2024)]:
    totals = data.loc[
        data["state_occurrence"] == travel_summary_row,
        residence_columns
    ].iloc[0]

    travel_comparison[f"travelers_{year}"] = pd.to_numeric(
        totals,
        errors="coerce"
    ).to_numpy()

travel_comparison["state_name"] = (
    travel_comparison["state"].map(state_names)
)

travel_comparison["change_in_travelers"] = (
    travel_comparison["travelers_2024"] -
    travel_comparison["travelers_2023"]
)

travel_comparison.sort_values(
    "change_in_travelers",
    ascending=False
)[
    [
        "state_name",
        "travelers_2023",
        "travelers_2024",
        "change_in_travelers"
    ]
]
```

## Geographic Distribution of Abortion Provision

```python
# Calculate each state's share of recorded provision across the 50 states and DC.

provision_shares = abortion_before.loc[abortion_before["geography"].isin(state_names.values()),
    ["geography", "abortions_2020"].copy()]

provision_shares = provision_shares.rename(columns={"geography": "state_name"})

provision_shares = provision_shares.merge(annual_2024[["state_name", "abortions_2024"]],
    on="state_name",
    how="left",
    validate="one_to_one")

print("States and DC:", len(provision_shares))
print("Missing values:")
print(provision_shares.isna().sum())


for year in [2020, 2024]:provision_shares[f"share_{year}"] = (
        provision_shares[f"abortions_{year}"] /
        provision_shares[f"abortions_{year}"].sum()) * 100

provision_shares["share_change_percentage_points"] = (provision_shares["share_2024"] - provision_shares["share_2020"])

provision_shares.sort_values("share_change_percentage_points", ascending=False)[
    ["state_name", "share_2020", "share_2024", "share_change_percentage_points"]].round(2)
```

```python
share_changes = provision_shares.sort_values(
    "share_change_percentage_points"
)

colors = [
    "steelblue" if change >= 0 else "darkorange"
    for change in share_changes["share_change_percentage_points"]
]

plt.figure(figsize=(10, 14))

plt.barh(
    share_changes["state_name"],
    share_changes["share_change_percentage_points"],
    color=colors
)

plt.axvline(0, color="black", linewidth=1)

plt.title("Change in State Shares of Abortion Provision, 2020–2024")
plt.xlabel("Change in share (percentage points)")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

### Results Shown

## Exploratory Statistical Inference

```python
# Adjust travel counts for population size and keep states with rates available in both years.

travel_rates = travel_comparison.merge(
    population_23,
    on="state_name",
    how="left",
    validate="one_to_one"
)

travel_rates = travel_rates.merge(
    population_24,
    on="state_name",
    how="left",
    validate="one_to_one"
)

travel_rates["travel_rate_2023"] = (
    travel_rates["travelers_2023"] /
    travel_rates["women_15_44_2023"]
) * 1000

travel_rates["travel_rate_2024"] = (
    travel_rates["travelers_2024"] /
    travel_rates["women_15_44_2024"]
) * 1000

travel_pairs = travel_rates.dropna(
    subset=["travel_rate_2023", "travel_rate_2024"]
).copy()

travel_pairs["rate_change"] = (
    travel_pairs["travel_rate_2024"] -
    travel_pairs["travel_rate_2023"]
)

print("States included:", len(travel_pairs))

sns.histplot(data=travel_pairs, x="rate_change", bins=12)
plt.axvline(0, color="black", linestyle="--")
plt.title("Changes in Interstate Abortion Travel Rates")
plt.xlabel("2024 minus 2023: travelers per 1,000 women aged 15–44")
plt.ylabel("Number of states")
plt.tight_layout()
plt.show()
```

```python
# Compare each state's 2023 and 2024 travel rates using a two-sided paired t-test.
from scipy import stats

alpha = 0.05

travel_test = stats.ttest_rel(
    travel_pairs["travel_rate_2024"],
    travel_pairs["travel_rate_2023"]
)

print("Mean state travel rate in 2023:",
      round(travel_pairs["travel_rate_2023"].mean(), 3))

print("Mean state travel rate in 2024:",
      round(travel_pairs["travel_rate_2024"].mean(), 3))

print("Mean change:",
      round(travel_pairs["rate_change"].mean(), 3))

print("t-statistic:", round(travel_test.statistic, 3))
print("Degrees of freedom:", len(travel_pairs) - 1)
print("P-value:", round(travel_test.pvalue, 4))
print("Alpha:", alpha)

if travel_test.pvalue < alpha:
    print("Reject H_0: There is statistically significant evidence of a change in the mean state interstate abortion travel rate between 2023 and 2024, under the test's assumptions."
    )
else:
    print("Fail to reject H_0: There is insufficient evidence of a change in the mean state interstate abortion travel rate between 2023 and 2024."
    )
```

### Results
As the mean travel rate across 47 states decreased by 0.345 travellers per 1000 women between the ages of 15-44. 
With p = 0.0208, the paired t-test indicates a statistically significant change at α = 0.05, under its assumptions.

```python
# Compare monthly averages for the same calendar periods in 2023 and 2024.

florida_monthly["period"] = np.where(
    florida_monthly["month_number"] <= 4,
    "January–April",
    "May–December"
)

florida_summary = (
    florida_monthly.groupby(["period", "year"])
    .agg(
        months_available=("month_number", "nunique"),
        average_monthly_abortions=("median", "mean")
    )
    .reset_index()
)

print(florida_summary.round(1).to_string(index=False))

period_comparison = florida_summary.pivot(
    index="period",
    columns="year",
    values="average_monthly_abortions"
)

period_comparison.columns = [
    "monthly_average_2023",
    "monthly_average_2024"
]

period_comparison["change"] = (
    period_comparison["monthly_average_2024"] -
    period_comparison["monthly_average_2023"]
)

period_comparison["percent_change"] = (
    period_comparison["change"] /
    period_comparison["monthly_average_2023"]
) * 100

period_comparison.round(2)
```
